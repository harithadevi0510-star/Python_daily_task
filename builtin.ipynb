# List Comprehensions

```python
n=[i for i in range(1,11)]
print(n)
```

```python
# squares of numbers
n1=[i*i for i in range(1,11)]
print(n1)
```

```python
# even numbers
n2=[i for i in range(1,21) if i%2==0]
print(n2)
```

```python
result=[i*3 for i in range(10)]
print(result)
```

```python
result=[x for x in range(5)]
print(result)
```

```python
result=[i*7 for i in range(6)]
print(result)
```

```python
words=["akshi","varsh","hari"]
upper_word=[word.upper() for word in words]
print(upper_word)
```

```python
result=[i**2 for i in range(1,6)]
print(result)
```

```python
result= [i for i in range(0,10)]
print(result)
```

```python
numbers=[-2,-1,0,1,2,3]
positive=[i for i in numbers if i>0]
print(positive)
```

```python
# List Comprehension with conditions
result=[i for i in range(1,31) if i%3==0]
print(result)
```

```python
result=[i for i in range(1,16) if i%2!=0]
print(result)
```

```python
num=["even" if i%2==0 else "odd" for i in range(1,9)]
print(num)
```

```python
result=[i*i for i in range(1,11) if i%2==0]
print(result)
```

```python
result=[x for x in range(10) if x%2==0]
print(result)
```

```python
n="Python Programming"
result=[ch for ch in n if ch.lower() not in "aeiou"]
print(result)
```

```python
a=["cat","apple","banana","grapes"]
result=[x for x in a if len(x)>4]
print(result)
```

```python
result=[(i,i**2) for i in range(5)]
print(result)
```

```python
result=[x if x>5 else 0 for x in range(10)]
print(result)
```

```python
# Advanced and Nested List Comprehensions
table=[[i*j for j in range(1,6)]
for i in range(1,6)]
print(table)
```

```python
a=[[1,2],[3,4],[5,6]]
print(a)
```

```python
char=["varsh","akshi","hari"]
res=[ch for char in char for ch in char]
print(res)
```

```python
matrix=[[i+j*3+1 for i in range(3)] for j in range(3)]
print(matrix)
```

```python
names=["aksh","hari","varsh"]
result=[(i,name) for i,name in enumerate(names)]
print(result)
```

```python
list1=[1,2,3,4,5,6]
list2=[4,5,6,7,8,9]
r=[i for i in list1 if i in list2]
print(r)
```

```python
import math
result=[math.factorial(i) for i in range(1,7)]
print(result)
```

```python
result=[[j for j in range(i+1)] for i in range(4)]
print(result)
```
