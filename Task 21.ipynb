```python
# parent class:
class Person:
    def __init__(self, name, age):
        self.name=name
        self.age=age

#child class:1
class Student(Person):
    def __init__(self, name, age, marks):
        super().__init__(name, age)
        self.marks=marks

   #display info
    def display(self):
        print("name:", self.name)
        print("age:", self.age)
        print("marks:", self.marks)

#child class:2
class Teacher(Person):
    def __init__(self, name, age, subject):
        super().__init__(name, age)
        self.subject=subject

    # display info
    def display(self):
        print("name:", self.name)
        print("age:", self.age)
        print("marks:", self.subject)

s = Student("Haritha",22,85)
s.display()
t=Teacher("Karunya",32,"Maths")
t.display()
```

```python
#parent class
class Shape:
    def area(self):
        pass

#child class:1
class Circle(Shape):
    def __init__(self, radius):
        self.radius=radius

    def area(self):
        print("area of circle:", 3.14 * self.radius * self.radius)
#child class :2
class Rectangle(Shape):
    def __init__(self, length, breadth):
        self.length=length
        self.breadth=breadth

    def area(self):
        print("area of rectangle:", self.length * self.breadth)

c=Circle(5)
c.area()
r=Rectangle(4,6)
r.area()
        
```

```python
# parent class
class BankAccount:
    def __init__(self, balance):
        self.balance = balance
    def deposit(self, amount):
        self.balance += amount

#child class:1
class SavingsAccount(BankAccount):
    def deposit(self, amount):
        self.balance += amount
        print(f"SavingsAccount:{self.balance}")

#child class:2
class CurrentAccount(BankAccount):
    def deposit(self, amount):
        self.balance += amount
        print(f"CurrentAccount:{self.balance}")

#object
s=SavingsAccount(40000)
c=CurrentAccount(50000)

#Calling inheritance
s.deposit(2000)
c.deposit(4000)

#Calling inheritance
SavingsAccount : 42000
CurrentAccount : 52000

         
    
```

```python
# parent class
class Book:
    def __init__(self, title, author):
        self.title=title
        self.author=author

#child class:1
class fiction(Book):
    def __init__(self, title, author):
        super().__init__(title, author)

# display info
    def display(self):
        print("fiction book title:", self.title)
        print("fiction book author:", self.author)

#child class:2
class nonfiction(Book):
    def __init__(self, title, author):
        super().__init__(title, author)

# display info
    def display(self):
        print("nonfiction book title:", self.title)
        print("nonfiction book author:", self.author)

#create object
f=fiction("Harry Potter","J.K.Rowling")
n=nonfiction("Wings of Fire","Atomic Habits")
f.display()
n.display()
```

```python
#parent class
class Account:
    def __init__(self, balance):
        self.balance=balance
        
    def show_balance(self):
        print("Balance:", self.balance)

#child class:1
class UserAccount(Account):
    def __init__(self, balance, username):
        super().__init__(balance)
        self.username=username
        
    def display(self):
        print("User Name:", self.username)
        self.show_balance()

#child class:2
class AdminAccount(Account):
    def __init__(self, balance, adminname):
        super().__init__(balance)
        self.adminname=adminname
        
    def display(self):
        print("Admin Name:", self.adminname)
        self.show_balance()

#create object
u=UserAccount(30000, "Haritha")
a=AdminAccount(40000, "Varshini")

#call methods
u.display()
a.display()
    
        
```

```python
#parent class
class BankAccount:
    def __init__(self, balance):
        self.balance = balance
#child class:1
class Savings(BankAccount):
    def __init__(self,balance):
        super().__init__(balance)

    def interest(self):
        amt=self.balance+(self.balance*5/100)
        print(f"Savings:{self.balance}")
        print(f"Balance after interest:",amt)

#child class:2
class Current(BankAccount):
    def __init__(self, balance):
        super().__init__(balance)

    def interest(self):
        amt=self.balance+(self.balance*2/100)
        print(f"Current:{self.balance}")
        print(f"Balance after interest:",amt)

#object
s=Savings(45000)
c=Current(56000)

#Calling inheritance
s.interest()
c.interest()
        

```

```python
# parent class
class Account:
    def __init__(self, balance):
        self.balance=balance

#child class:1
class Savings(Account):
    def __init__(self, balance, interest):
        super().__init__(balance)
        self.interest=interest

    def display(self):
        print(f"Savings:{self.balance}")
        print(f"interest:{self.interest}")
       
#child class:2
class Current(Account):
    def __init__(self, balance, overdraft):
        super().__init__(balance)
        self.overdraft=overdraft

    def display(self):
        print(f"Current:{self.balance}")
        print(f"Overdraft:{self.overdraft}")

#child class:3
class FixedDeposit(Account):
    def __init__(self, balance, maturity):
        super().__init__(balance)
        self.maturity=maturity

    def display(self):
        print(f"FixedDeposit:{self.balance}")
        print(f"maturity:{self.maturity}")
#object
s=Savings(45000,4)
c=Current(60000,1000)
f=FixedDeposit(20000,3)
#calling inheritance
s.display()
c.display()
f.display()

       
```

```python
#parent class
class Employee:
    def __init__(self, name):
        self.name = name
    def calculate_salary(self):
        print("Salary calculation")
#child class:1
class FullTime(Employee):
    def __init__(self, name, monthly_salary):
        super().__init__(name)
        self.monthly_salary = monthly_salary
    def calculate_salary(self):
        print("Employee:", self.name)
        print("Monthly Salary:", self.monthly_salary)
#child class:2
class PartTime(Employee):
    def __init__(self, name, hours, hourly_rate):
        super().__init__(name)
        self.hours = hours
        self.hourly_rate = hourly_rate
    def calculate_salary(self):
        salary = self.hours * self.hourly_rate
        print("Employee:", self.name)
        print("Hourly Salary:", salary)
#create object
f = FullTime("Aks", 50000)
p = PartTime("kams", 80, 500)
# call the object
f.calculate_salary()
p.calculate_salary()
```

```python
#parent class
class Animal:
    def __init__(self, name, breed):
        self.name = name
        self.breed = breed
        
    def sound(self):
        print(self.name, "makes a sound")

#child class:1
class Cat(Animal):
    def meow(self):
        print("Cat meows")

#child class:2
class Dog(Animal):
    def bark(self):
        print("Dog barks")

#child class:3
class Cow(Animal):
    def moo(self):
        print("Cow says Moo")

#create object
c = Cat("Tommy", "Labrador")
d = Dog("Bunny", "Persian")
co = Cow("Ganga", "Gir")
# call the object
print("Name:", c.name)
print("Breed:", c.breed)
c.sound()
c.meow()
print("Name:", d.name)
print("Breed:", d.breed)
d.sound()
d.bark()
print("Name:", co.name)
print("Breed:", co.breed)
co.sound()
co.moo()
```
