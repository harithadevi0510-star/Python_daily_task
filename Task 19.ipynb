# 6

```python
#parent class
class BankAccount:
    #constructor
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print("Deposited:", amount)

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Withdraw:", amount)
        else:
            print("Insufficient Balance")

#child class
class SavingsAccount(BankAccount):
    def __init__(self, balance):
        super().__init__(balance)

    def calculate_interest(self):
        interest = self.balance*0.05
        print("Interest:", interest)

s = SavingsAccount(20000)
print("Initial Balance:", s.balance)
s.deposit(3000)
print("Balance after deposit:", s.balance)
s.withdraw(2500)
print("Balance after withdrawl:", s.balance)

s.calculate_interest()
            
       
        
```

```python
# grand parent
class BankAccount:
    #constructor
    def __init__(self, balance):
        self.balance = balance
        
    def deposit(self, amount):
        self.balance += amount
        print("Deposited:", amount)
        
# parent class
class Savings_Account(BankAccount):
    #constructor
    def __init__(self, balance, rate):
        super().__init__(balance)
        self.rate = rate
        
    def calculate_interest(self):
        interest = (self.balance * self.rate) / 100  # interest=(bal*rate)/100
        print("Interest:", interest)

# child class
class Recurring_Deposit(Savings_Account):
    # constructor
    def __init__(self, balance, rate, months, monthly_deposit):
        super().__init__(balance, rate)
        self.months = months
        self.monthly_deposit = monthly_deposit
        
    def calculate_maturity(self):   #maturity amount=total deposit + interest
        total_deposit = self.months * self.monthly_deposit
        interest = (total_deposit * self.rate) / 100
        maturity_amount = total_deposit + interest
        print("Total Deposit:", total_deposit)
        print("Maturity Amount:", maturity_amount)
        
r = Recurring_Deposit(5000, 5, 12, 1000)
r.deposit(2000)
r.calculate_interest()
r.calculate_maturity()
     
    
```

```python
#grandparent
class Account:
    def __init__(self, balance):
        self.balance = balance
        
    def calculate_interest(self):
        interest = (self.balance * 2) / 100
        print("Interest:", interest)

#parent class
class Saving_Account(Account):
    def calculate_interest(self):
        interest = (self.balance * 3) / 100
        print("Interest:", interest)

#child class
class Senior_CitizenAccount(Saving_Account):
    def calculate_interest(self):
        interest = (self.balance * 4) / 100
        print("Interest", interest)
        
a=Account(10000)
s = Saving_Account(10000)
sc = Senior_CitizenAccount(10000)

a.calculate_interest()
s.calculate_interest()
sc.calculate_interest()
```
