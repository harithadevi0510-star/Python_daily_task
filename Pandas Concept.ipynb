# OOP + Encapsulation

```python
# Problem 10:
class Student:
    def __init__(self, marks):
        self.__marks=0

    def get_marks(self):
        return self.__marks

    def set_marks(self, marks):
        if 0 <= marks <= 100:
            self.__marks=marks
        else:
            print("Marks  should be between 0 and 100")
            
s=Student(55)
s.set_marks(85)
print("Marks:", s.get_marks())           
```

```python
# Problem 11:
class InsufficientBalanceError(Exception):
    pass
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        self.__balance += amount
        print("Deposited:", amount)
        print("Balance:", self.__balance)

    def withdraw(self, amount):
        if self.__balance - amount < 0:
            raise InsufficientBalanceError("Balance cannot be negative.")
        self.__balance -= amount
        print("Withdrawn:", amount)
        print("Balance:", self.__balance)

account = BankAccount(5000)
account.deposit(1000)
try:
    account.withdraw(2000)
    account.withdraw(5000)
except InsufficientBalanceError as e:
    print(e)
```

```python
# Problem 13:
class Product:
    def __init__(self, price):
        self.__price = 200
        self.set_price(price)

    def get_price(self):
        return self.__price

    def set_price(self, price):
        if price >= 200:
            self.__price = price
        else:
            print("Price must be at least 200.")
p = Product(800)
print("Price:", p.get_price())
p.set_price(900)
print("Updated Price:", p.get_price())
p.set_price(160)
print("Final Price:", p.get_price())
```

```python
# Problem 14:
class Rectangle:
    def __init__(self, length, width):
        self.__length = length
        self.__width = width
    def area(self):
        return self.__length * self.__width

    def perimeter(self):
        return 2 * (self.__length + self.__width)
r = Rectangle(2, 4)
print(r.area())
print(r.perimeter())
```

```python
# Problem 15:
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
class Student(Person):
    def __init__(self, name, age, course):
        super().__init__(name, age)   
        self.course = course
    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)
        print("Course:", self.course)
s = Student("Haritha", 22, "DS")
s.display()
```

```python
#Problem 18:
from abc import ABC, abstractmethod
class Shape(ABC):
    def __init__(self):
        print("Shape Constructor")
    @abstractmethod
    def area(self):
        pass
class Circle(Shape):
    def __init__(self, radius):
        super().__init__()   
        self.radius = radius
    def area(self):
        return 3.14 * self.radius * self.radius
class Square(Shape):
    def __init__(self, side):
        super().__init__()   
        self.side = side
    def area(self):
        return self.side * self.side
c = Circle(5)
s = Square(4)
print( c.area())
print(s.area())
```

```python
# Problem 19
class Father:
    def show(self):
        print("Father")
class Mother:
    def show(self):
        print("Mother")
class Child(Father, Mother):
    def show(self):
        print("Method conflict resolved:")
        Father.show(self)      
        Mother.show(self)      

c = Child()
c.show()
```
