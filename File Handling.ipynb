```python
#pickle file  # adding for the dump  # reading for the load

import pickle

students = {"Name":"Haritha",
            "Age":22,
            "Mob-No":12345
           }
with open("pickle.bin","wb") as file:
    pickle.dump(students,file)

print("File write Success")
            
```

```python
#read the binary file

with open("pickle.bin","rb") as file:
    r=pickle.load(file)
    print(r)

print("File read Success")
```

# 29

```python
while True:
    print("1.Write")
    print("2.Read")
    print("3.Append")
    print("4.Exit")
    choice = input()
    if choice == "1":
        file = open("data.txt", "w")
        file.write(input())
        file.close()
        print("Written")
    elif choice == "2":
        file = open("data.txt", "r")
        print(file.read())
        file.close()
    elif choice == "3":
        file = open("data.txt", "a")
        file.write(input())
        file.close()
    elif choice == "4":
        break
    else:
        print("invalid choice")
```
