# args(Multiple Positional Arguments)

```python
def add_all(*args):
    c=sum(args)
    return c
add_all(1,2,3,4,5,6,7,8,9)
```

```python
def print_all(*args):
    return args
print(print_all("hari", "unnu", "vivansh"))
    
```

```python
def find_max(*args):
    return max(args)
result=find_max(50, 45, 33,90)
print(result)
```

```python
def concatenate(*args):
    return " ".join(args)
result = concatenate("Hi","Hello","Morning")
print(result)
    
```

```python
def multiply(*args):
    c = 1
    for num in args:
        c = c*num
    return c
print(multiply(1,2,3,4,5,6,7,8,9))
```

```python
def average(*args):
    return sum(args) / len(args)
print(average(10,20,30,40,50))
```

```python
def info(*args):
    for item in args:
        print("Value:",item)
        print("Type:",type(item))
info(22, "Biscuit", 12.5, True)
    
```

# kwargs (Multiple Keyword Arguments)

```python
def student_info(**kwargs):
    for key, value in kwargs.items():
        print(key, ":", value)
student_info(name="Hari", age=22, course="Python")
    
```

```python
def create_profile(**kwargs):
    for key, value in kwargs.items():
        print(key, ":", value)
create_profile(name="Hari",age=22, city="Warangal")
    
```
