```python
employees = [
    {"name":"Ram","age":25,"salary":50000},
    {"name":"Amit","age":25,"salary":60000},
    {"name":"Neha","age":23,"salary":45000},
    {"name":"Raj","age":27,"salary":55000}
]
result = sorted(employees, key=lambda x: x["age"])
result1 = sorted(employees, key=lambda x: -x["salary"])
print("Sorted by Age:")
print(result)
print("\nSorted by Salary (Descending):")
print(result1)
```
