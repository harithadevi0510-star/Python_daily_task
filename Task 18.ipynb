```python
from abc import ABC, abstractmethod
class Shape(ABC) :
    # constructor
    def __init__(self,radius,length,breadth):
        self.__radius=radius
        self.__length=length
        self.__breadth=breadth
        
 # abstract method
    @abstractmethod
    def area(self):
        pass
    @abstractmethod
    def perimeter(self):
        pass
# child class:1       
class Circle(Shape):
    def __init__(self,radius):
        self.radius=radius
    def area(self):
        return 3.14*self.radius*self.radius
        
    def perimeter(self):
        return 2*3.14*self.radius
# child class:2
class Rectangle(Shape):
    def __init__(self,length,breadth):
        self.length=length
        self.breadth=breadth
        
    def area(self):
        return self.length*self.breadth
        
    def perimeter(self):
        return 2*(self.length+self.breadth)
c=Circle(10)
print("Circle area:",c.area())
print("Circle perimeter:",c.perimeter())
r=Rectangle(20,5)
print("rectangle area:",r.area())
print("rectangle perimeter:",r.perimeter())

    
```

```python
from abc import ABC, abstractmethod
class Shape3D(ABC):
    #abstract method
    @abstractmethod
    def volume(self):
        pass
    #abstract method
    @abstractmethod
    def surface_area(self):
        pass
#child class:1
class Cube(Shape3D):
    def __init__(self, side):
        self.side = side

    def volume(self):
        return self.side ** 3

    def surface_area(self):
        return 6 * (self.side ** 2)
#child class:2
class Sphere(Shape3D):
    def __init__(self, radius):
        self.radius = radius

    def volume(self):
        return (4/3) * 3.14 * (self.radius ** 3)

    def surface_area(self):
        return 4 * 3.14 * (self.radius ** 2)

c = Cube(6)
s = Sphere(4)

print("Cube Volume:", c.volume())
print("Cube Surface Area:", c.surface_area())
print("Sphere Volume:", s.volume())
print("Sphere Surface Area:", s.surface_area())

```

```python
from abc import ABC, abstractmethod

class Report(ABC):
    
    # constructor
    def __init__(self,title):
        self.title=title
    def show_header(self):
        print("Report Title:",self.title)
    
    # abstact method
    @abstractmethod
    def generate(self):
        pass
        
#child class:1
class PDFReport(Report):
    def __init__(self, title):
        super().__init__(title)
        
    def generate(self):
        super().show_header()
        print("PDFReport is generated")
        
#child class :2
class HTMLReport(Report):
    def __init__(self,title,filename):
        super().__init__(title)
        self.filename=filename
        
    def generate(self):
        super().show_header()
        print("HTMLReport is generated")


p = PDFReport("Student Report", "student_report.pdf")
print("Title    :", p.title)
print("Filename :", p.filename)
p.generate() 

print()

h = HTMLReport("Sales Report", "sales_report.html")
print("Title    :", h.title)
print("Filename :", h.filename)

h.generate()

```
