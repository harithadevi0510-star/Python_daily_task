```python
# grand parent class
class Person:
    
    #constructor
    def __init__(self,name):
        self.name=name
    # method
    def person_info(self):
        print(f"Person name:{self.name}")

# parent class
class Employee(Person):
    
    #constructor
    def __init__(self, name, emp_id):
        super().__init__(name)
        self.emp_id=emp_id
 # method
    def emp_info(self):
        print(f"employee id: {self.emp_id}")

#child class
class Manager(Employee):
    
    #constructor
    def __init__(self,name,emp_id,department):
        super().__init__(name,emp_id)
        self.department=department

    # method
    def Manager_info(self):
        print(f"department: {self.department}")
M=Manager("Haritha",426,"Data Science")
M.person_info()
M.emp_info()
M.Manager_info()
```

```python
# parent class :1
class Father:
    #constructor
    def __init__(self, father_name, **kwargs):
        super().__init__(**kwargs)
        self.father_name = father_name

# parent class :2
class Mother:
    #constructor
    def __init__(self, mother_name, **kwargs):
        super().__init__(**kwargs)
        self.mother_name = mother_name
        
#child class
class Child(Father, Mother):
    #constructor
    def __init__(self, father_name, mother_name, child_name):
            super().__init__(
                father_name=father_name,
                mother_name=mother_name
            )
            self.child_name = child_name
        
# display method
    def display(self):
        print("Father :", self.father_name)
        print("Mother :", self.mother_name)
        print("Child :", self.child_name)



c = Child("Nagaraju", "SandhyaRani", "Haritha Devi")
c.display()
```

# 23

```python
from abc import ABC, abstractmethod
#parent class
class Shape(ABC):
    def __init__(self):
        print("Shape")
    #abstract method
    @abstractmethod
    def area(self):
        pass
        
#child class:1
class Circle(Shape):
    def __init__(self,radius):
        super().__init__()
        self.radius=radius
    def area(self):
        print(3.14*self.radius*self.radius)
        
#child class:2
class Rectangle(Shape):
    def __init__(self,length,width):
        super().__init__()
        self.length=length
        self.width=width
    def area(self):
        print(self.length*self.width)

c=Circle(6)
c.area()
r=Rectangle(2,3)
r.area()
```

```python
from abc import ABC, abstractmethod
class Shape(ABC):
    total_shapes = 0  
    #constructor
    def __init__(self):
        Shape.total_shapes += 1
    #abstract method
    @abstractmethod
    def area(self):
        pass
    #abstract method
    @abstractmethod
    def perimeter(self):
        pass
    #class method
    @classmethod
    def show_total_shapes(cls):
        print("Total Shapes Created:", cls.total_shapes)
#child class:1        
class Circle(Shape):
    #constructor
    def __init__(self, radius):
        super().__init__()      
        self.radius = radius
    def area(self):
        return 3.14 * self.radius * self.radius
    def perimeter(self):
        return 2 * 3.14 * self.radius
#child class:2
class Rectangle(Shape):
    def __init__(self, length, width):
        super().__init__()      
        self.length = length
        self.width = width
    def area(self):
        return self.length * self.width
    def perimeter(self):
        return 2 * (self.length + self.width)
#create object
c = Circle(5)
r = Rectangle(10, 4)
print("Circle Area:", c.area())
print("Circle Perimeter:", c.perimeter())
print("Rectangle Area:", r.area())
print("Rectangle Perimeter:", r.perimeter())
Shape.show_total_shapes()
```

```python
# Parent class
class Person:
    #constructor
    def __init__(self, name, age):
        self.__name = name     
        self.__age = age
     #Getter
    def get_name(self):
        return self.__name
    def get_age(self):
        return self.__age
    #setter
    def set_name(self, name):
        self.__name = name
    def set_age(self, age):
        if age > 0:
            self.__age = age
        else:
            print("Age must be greater than 0")
    #display info
    def display(self):
        print("Name:", self.__name)
        print("Age:", self.__age)
#child class:1
class Employee(Person):
    #constructor
    def __init__(self, name, age, emp_id, salary):
        super().__init__(name, age)
        self.__emp_id = emp_id     
        self.__salary = salary
    #Getter
    def get_emp_id(self):
        return self.__emp_id
    def get_salary(self):
        return self.__salary
    #setter
    def set_emp_id(self, emp_id):
        self.__emp_id = emp_id
    def set_salary(self, salary):
        if salary >= 5000:
            self.__salary = salary
        else:
            print("Salary must be at least 5000")
        #display info
    def display(self):
        super().display()
        print("Employee ID:", self.__emp_id)
        print("Salary:", self.__salary)
#child class :2
class Manager(Employee):
    #constructor
    def __init__(self, name, age, emp_id, salary, department):
        super().__init__(name, age, emp_id, salary)
        self.__department = department  
    #Getter
    def get_department(self):
        return self.__department
    #Setter
    def set_department(self, department):
        self.__department = department
    #display info
    def display(self):
        super().display()
        print("Department:", self.__department)

m = Manager("Haritha", 23, 101, 50000, "Developer")
m.display()
m.set_age(23)
m.set_salary(55000)
m.set_department("Finance")
print("\nAfter Updating Details:")
m.display()
```

```python
#parent class:1
class Camera:
    def feature(self):
        print("Camera: Takes high-quality photos")
#parent class:2
class MusicPlayer:
    def feature(self):
        print("MusicPlayer: Plays music")
#child class
class Smartphone(Camera, MusicPlayer):
    def feature(self):
        print("Smartphone Features:")
        super().feature()   
    def show_mro(self):
        print("Method Resolution Order (MRO):")
        print(Smartphone.mro())
#create method       
phone = Smartphone()
#call the method
phone.feature()
phone.show_mro()
```
