# Level 2: Intermediate Iterator Problems

```python
#Problem 6
class NumberIterator:
    def __init__(self):
        self.current = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.current <= 10:
            num = self.current
            self.current += 1
            return num
        else:
             raise StopIteration
       
numbers = NumberIterator()

for num in numbers:
    print(num)
```

```python
class RevOfNumbers:
    def __init__(self):
        self.current = 10

    def __iter__(self):
        return self

    def __next__(self):
        if self.current >= 1:
            value = self.current
            self.current -= 1
            return value
        else:
             raise StopIteration
       
numbers = RevOfNumbers()

for num in numbers:
    print(num)
```

```python
# Problem 9
class WordIterator:
    def __init__(self,word):
        self.word=word
        self.index=0
    
    def __iter__(self):
        return self

    def __next__(self):
        if self.index < len(self.word):
            ch=self.word[self.index]
            self.index += 1
            return ch
        else:
            raise StopIteration
word=WordIterator("GENERATOR")
for ch in word:
    print(ch)
```

# Level 4: Generator with Lists

```python
# Problem 16
numbers=[10, 15, 20, 25, 30, 35]
def even_numbers(num):
    for i in num:
        if i % 2 == 0:
            yield i
            
gen = even_numbers(numbers)

for num in gen:
    print(num)
```

```python
# Problem 17
numbers=[1, 4, 7, 10, 13, 16, 19]
def odd_numbers(num):
    for i in num:
        if i % 2 != 0:
            yield i
            
gen = odd_numbers(numbers)

for num in gen:
    print(num)
```

```python
# Problem 18
names = ["Alice", "Bob", "Charlie", "David", "Emma"]
def name_generator(names):
    for name in names:
        yield name
        
gen = name_generator(names)

for name in gen:
    print(name)
```

```python
# Problem 19
numbers = [20, 55, 40, 90, 60, 15, 75]

def greater(nums):
    for num in nums:
        if num > 50:
            yield num

gen = greater(numbers)

for num in gen:
    print(num)
```

```python
## Problem 21
class NaturalNumber:
    def __init__(self):
        self.num=1
    def __iter__(self):
        return self
    def __next__(self):
        if self.num<=15:
            value=self.num
            self.num+=1
            return value
        else:
            raise StopIteration
obj=NaturalNumber()
for i in obj:
    print(i)
```

```python
# Problem 23
class Alphabet:
    def __init__(self):
        self.letter=ord('A')
    def __iter__(self):
        return self
    def __next__(self):
        if self.letter <= ord('Z'):
            value=chr(self.letter)
            self.letter+=1
            return value
        else:
            raise StopIteration
obj=Alphabet()
for letter in obj:
    print(letter)
```

```python
# Problem 24
class Vowel:
    def __init__(self, text):
        self.text = text
        self.index = 0
    def __iter__(self):
        return self
    def __next__(self):
        vowels = "AEIOUaeiou"
        while self.index < len(self.text):
            char = self.text[self.index]
            self.index += 1
            
            if char in vowels:
                return char

        raise StopIteration
text = "Artificial Intelligence"
vowels = Vowel(text)

for v in vowels:
    print(v)
```

```python
# Problem 25
class Fibonacci:
    def __init__(self):
        self.a=0
        self.b=1
        self.count=0
    def __iter__(self):
        return self
    def __next__(self):
        if self.count<10:
            value=self.a
            self.a, self.b=self.b, self.a+self.b
            self.count += 1
            return value
        else:
            raise StopIteration
t=Fibonacci()
for num in t:
    print(num)
```

# Level 6: Advanced Generator Problems

```python
# Problem 26
def fibonacci():
    a,b=0,1
    for i in range(15):
        yield a
        a,b=b, a+b
for num in fibonacci():
    print(num)
```

```python
#Problem 27
def prime_numbers():
    for num in range(2, 101):
        is_prime = True
        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                is_prime = False
                break
        if is_prime:
            yield num
for prime in prime_numbers():
    print(prime)
```

```python
# Problem 28
def factors(num):
    for i in range(1, num + 1):
        if num % i == 0:
            yield i
num = 24
for factor in factors(num):
    print(factor)
```

```python
# Problem 29
def countdown():
    for i in range(20, 0, -1):
        yield i
for num in countdown():
    print(num)
```

```python
# Problem 30
def multiples_of_five():
    num = 5
    while True:
        yield num
        num += 5
gen = multiples_of_five()
for i in range(15):
    print(next(gen))
```
